### Transformation Phase

# orders 

```python
orderDf.head()

# calculating the days it took to deliver the order
orderDf["delivery_days"] = (orderDf["order_delivered_customer_date"]- orderDf["order_purchase_timestamp"]).dt.days

orderDf[
    [
        "order_purchase_timestamp",
        "order_delivered_customer_date",
        "delivery_days"
    ]
].head(10)

# finding the year in which order was placed
orderDf["order_year"] = orderDf["order_purchase_timestamp"].dt.year

# finding out the month in which order was placed
orderDf["order_month"] = orderDf["order_purchase_timestamp"].dt.month

# finding out the weekday in which order was placed
orderDf["order_weekday"] = orderDf["order_purchase_timestamp"].dt.day_name()

# How many days earlier or later was the order delivered compared with the estimated delivery date?

orderDf["delivery_vs_estimated_days"] = (
    orderDf["order_delivered_customer_date"]
    - orderDf["order_estimated_delivery_date"]
).dt.days

orderDf[
    [
        "order_delivered_customer_date",
        "order_estimated_delivery_date",
        "delivery_vs_estimated_days"
    ]
].head(10)
```
